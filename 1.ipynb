```python
def check_time(hour):
    h,m = hour.split(":")
    if datetime.now().hour == int(h) and (int(m) >= datetime.now().minute -5  or  int(m) <= datetime.now().minute + 5):
        return True

check_time(df.iloc[1]["utc"])
```

```python
for i,data in df.iterrows():

    time_format = "%H:%M"

    # Parse the times into datetime objects
    t1 = datetime.strptime(data["utc"], time_format)
    t2 = datetime.strptime(data["utc_final"], time_format)
    difference = t2 - t1
    print("Time difference:", int(difference.total_seconds()))
```

```python
def calculate_difference(data):
    time_format = "%H:%M"

    t1 = datetime.strptime(data["utc"], time_format)
    t2 = datetime.strptime(data["utc_final"], time_format)
    difference = t2 - t1
    return int(difference.total_seconds())
```
